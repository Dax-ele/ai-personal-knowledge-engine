```python
sentences = [
    "Python è un linguaggio molto usato per sviluppare applicazioni software",
    "Java viene utilizzato dagli sviluppatori per creare sistemi backend",
    "Gli algoritmi di machine learning imparano dai dati",
    "Il gatto dorme sul divano del salotto",
    "La pizza italiana è famosa in tutto il mondo"
]

embeddings = model.encode(sentences)

similarity = cosine_similarity(embeddings)

similarity

```

```python
documents = [
    "Python è un linguaggio molto usato per sviluppare applicazioni software",
    "Java viene utilizzato dagli sviluppatori per creare sistemi backend",
    "Gli algoritmi di machine learning imparano dai dati attraverso modelli matematici",
    "Il gatto dorme sul divano del salotto",
    "La pizza italiana è famosa in tutto il mondo"
]

document_embeddings = model.encode(documents)
```

```python
def search(query, top_k=3):
    
    # trasformiamo la domanda in embedding
    query_embedding = model.encode([query])
    
    # calcoliamo la similarità con tutti i documenti
    similarities = cosine_similarity(
        query_embedding,
        document_embeddings
    )[0]
    
    # prendiamo gli indici dei risultati migliori
    best_indexes = similarities.argsort()[::-1][:top_k]
    
    results = []
    
    for idx in best_indexes:
        results.append({
            "document": documents[idx],
            "score": similarities[idx]
        })
    
    return results
```

```python
search("Come posso sviluppare un'applicazione backend?")
```

```python
search("Come funziona il machine learning?")
```

```python
search("Dove mangiare una buona pizza?")
```
